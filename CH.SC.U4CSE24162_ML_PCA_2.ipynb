```python
import pandas as pd
import numpy as np

data = {
    'Variable': [
        'Climate', 'Housing', 'Health', 'Crime',
        'Transportation', 'Education', 'Arts',
        'Recreation', 'Economy'
    ],
    'PC1': [0.190, 0.544, 0.782, 0.365, 0.585, 0.394, 0.985, 0.520, 0.142],
    'PC2': [0.017, 0.020, -0.605, 0.294, 0.085, -0.273, 0.126, 0.402, 0.150],
    'PC3': [0.207, 0.204, 0.144, 0.585, 0.234, 0.027, -0.111, 0.519, 0.239]
}

df = pd.DataFrame(data)
df
```

```python
loadings = df[['PC1', 'PC2', 'PC3']]

variance = (loadings ** 2).sum()

print("Sum of squared loadings:")
print(variance.round(3))

print("\nVariance percentage:")
print((variance / variance.sum() * 100).round(1))
```

```python
highest_loading = df.loc[df['PC1'].idxmax()]

print("Variable with highest PC1 loading:", highest_loading['Variable'])
print("PC1 loading:", highest_loading['PC1'])
```
