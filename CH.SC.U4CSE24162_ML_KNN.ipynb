```python
import os

train_path = "machinelearninginaction3x/Ch02/trainingDigits"
test_path = "machinelearninginaction3x/Ch02/testDigits"

print("Training samples:", len(os.listdir(train_path)))
print("Test samples:", len(os.listdir(test_path)))
```

```python
def img2vector(filename):
    return_vect = np.zeros((1, 1024))

    with open(filename) as fr:
        for i in range(32):
            line = fr.readline()
            for j in range(32):
                return_vect[0, 32 * i + j] = int(line[j])

    return return_vect

print(img2vector(train_path + "/" + os.listdir(train_path)[0]).shape)
```

```python
training_files = os.listdir(train_path)

training_mat = np.zeros((len(training_files), 1024))
training_labels = []

for i, filename in enumerate(training_files):
    training_labels.append(int(filename.split('_')[0]))
    training_mat[i, :] = img2vector(train_path + "/" + filename)

print("Training data shape:", training_mat.shape)
print("Training labels:", len(training_labels))
```

```python
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(training_mat, training_labels)

print("KNN classifier trained successfully.")
```

```python
test_files = os.listdir(test_path)

correct = 0

for filename in test_files:
    true_label = int(filename.split('_')[0])
    test_vector = img2vector(test_path + "/" + filename)
    predicted_label = knn.predict(test_vector)[0]

    if predicted_label == true_label:
        correct += 1

accuracy = correct / len(test_files)

print("Test Accuracy:", accuracy)
```
