```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
```

```python
np.random.seed(42)

X = np.random.uniform(-5, 5, (200, 2))

y = (X[:, 1] > X[:, 0]**2 / 4 - 2).astype(int)

plt.scatter(X[:, 0], X[:, 1], c=y)
plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Random Quadratic Dataset")
plt.show()
```

```python
model = SVC(kernel='poly', degree=2)
model.fit(X, y)
```

```python
y_pred = model.predict(X)

print("Predictions:")
print(y_pred)
```
