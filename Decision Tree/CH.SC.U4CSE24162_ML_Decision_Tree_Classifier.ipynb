```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn import metrics
import seaborn as sn
import matplotlib.pyplot as plt
```

```python
data = {
    'Weather': ['Sunny', 'Sunny', 'Rainy', 'Rainy',
                'Sunny', 'Rainy', 'Sunny', 'Rainy',
                'Sunny', 'Rainy', 'Sunny', 'Rainy'],
    'Activity': ['Go out', 'Go out', 'Stay in', 'Stay in',
                 'Go out', 'Stay in', 'Go out', 'Stay in',
                 'Go out', 'Stay in', 'Go out', 'Stay in']
}

data = pd.DataFrame(data)

print(data)
```

```python
print(data.isnull().sum())
```

```python
data['Weather'] = data['Weather'].map({'Sunny': 1, 'Rainy': 0})
data['Activity'] = data['Activity'].map({'Go out': 1, 'Stay in': 0})

print(data)
```

```python
X = data[['Weather']]
y = data['Activity']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=5
)

print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

```python
model = DecisionTreeClassifier(
    criterion='entropy',
    random_state=5
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print("Predicted values:")
print(y_pred)
```

```python
conf_mat = metrics.confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(conf_mat)

accuracy_score = metrics.accuracy_score(y_test, y_pred)

print("Accuracy Score:", accuracy_score)
print("Accuracy in Percentage:", int(accuracy_score * 100), "%")
```

```python
conf_mat = pd.crosstab(
    y_test,
    y_pred,
    rownames=['Actual'],
    colnames=['Predicted']
)

sn.heatmap(conf_mat, annot=True)
plt.title("Decision Tree Classifier")
plt.show()
```

```python
print("Sunny:",
      "Go out" if model.predict(pd.DataFrame([[1]], columns=['Weather']))[0] == 1 else "Stay in")

print("Rainy:",
      "Go out" if model.predict(pd.DataFrame([[0]], columns=['Weather']))[0] == 1 else "Stay in")
```
